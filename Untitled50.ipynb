Data Cleaning & Preparation

```python
#Handling Missing Values
df = df.dropna(subset=['Price per Unit', 'Quantity'])
```

```python
#Removing Invalid Observations
df = df[(df['Price per Unit'] > 0) & (df['Quantity'] > 0)]
```

```python
#Creating Revenue Variable
df['Revenue'] = df['Price per Unit'] * df['Quantity']
```

Rows with missing or non-positive price and quantity values were removed to ensure meaningful regression results. Revenue was computed at the transaction level to support price optimization analysis.

# Exploratory Data Analysis (EDA)
Price vs Quantity Relationship

```python
plt.figure(figsize=(8,6))
sns.scatterplot(
    x='Price per Unit',
    y='Quantity',
    data=df,
    alpha=0.6
)
plt.title("Price vs Quantity Sold")
plt.xlabel("Price per Unit")
plt.ylabel("Quantity Sold")
plt.show()
```

The scatter plot indicates an inverse relationship between price and quantity sold, which aligns with economic demand theory.

### Preparing Data for Regression

```python
regression_df = df[['Price per Unit', 'Quantity']].copy()

regression_df.rename(
    columns={
        'Price per Unit': 'Price',
        'Quantity': 'Quantity_Sold'
    },
    inplace=True
)
```

## Linear Regression Model (Baseline Demand Model)
Linear regression is used to understand the absolute relationship between price and quantity sold.

    Quantity=β0​+β1​×Price+ϵ

```python
y_linear = regression_df['Quantity_Sold']
X_linear = regression_df[['Price']]

#Add intercept term
X_linear = sm.add_constant(X_linear)

#Fit linear regression model
linear_model = sm.OLS(y_linear, X_linear)
linear_results = linear_model.fit()

#Display results
print(linear_results.summary())
```

The negative price coefficient confirms that demand decreases as price increases

A statistically significant p-value indicates price is a meaningful predictor

 R-squared shows how much demand variation is explained by price alone



Linear regression measures absolute change, but pricing decisions require understanding percentage change. For this reason, a log–log regression model is used to estimate price elasticity of demand.

## Log–Log Regression (Elasticity Model)
    log(Quantity)=β0​+β1​log(Price)+ϵ

Here,
𝛽1
directly represents price elasticity of demand.

```python
regression_df['log_price'] = np.log(regression_df['Price'])
regression_df['log_quantity'] = np.log(regression_df['Quantity_Sold'])

regression_df = regression_df.dropna(subset=['log_price', 'log_quantity'])
```

```python
y_log = regression_df['log_quantity']
X_log = regression_df[['log_price']]

X_log = sm.add_constant(X_log)

elasticity_model = sm.OLS(y_log, X_log)
elasticity_results = elasticity_model.fit()

print(elasticity_results.summary())
```

->The coefficient of log_price is the price elasticity.

->A negative value indicates inverse price-demand relationship.

->If |elasticity| > 1 → demand is elastic.

## Regression Diagnostics
Residual Distribution

```python
residuals = elasticity_results.resid

sns.histplot(residuals, kde=True)
plt.title("Distribution of Regression Residuals")
plt.xlabel("Residuals")
plt.show()
```

### Residuals vs Fitted Values

```python
sns.scatterplot(
    x=elasticity_results.fittedvalues,
    y=residuals,
    alpha=0.5
)
plt.axhline(0, color='red', linestyle='--')
plt.xlabel("Fitted Values")
plt.ylabel("Residuals")
plt.title("Residuals vs Fitted Values")
plt.show()
```

## Price Optimization Using Elasticity Model

Simulate Price Range

```python
price_range = np.linspace(
    regression_df['Price'].min(),
    regression_df['Price'].max(),
    50
)
```

Predict Demand

```python
log_price_range = np.log(price_range)

prediction_df = pd.DataFrame({'log_price': log_price_range})
prediction_df = sm.add_constant(prediction_df)

predicted_log_quantity = elasticity_results.predict(prediction_df)
predicted_quantity = np.exp(predicted_log_quantity)
```

```python
#COMPUTE REVENUE
predicted_revenue = price_range * predicted_quantity
```

```python
optimal_price = price_range[np.argmax(predicted_revenue)]
optimal_price
```

## Revenue Optimization Visualization

```python
plt.figure(figsize=(8,5))
plt.plot(price_range, predicted_revenue, label='Estimated Revenue')
plt.axvline(
    optimal_price,
    color='red',
    linestyle='--',
    label=f'Optimal Price ≈ {optimal_price:.2f}'
)
plt.xlabel("Price per Unit")
plt.ylabel("Estimated Revenue")
plt.title("Revenue vs Price Optimization Curve")
plt.legend()
plt.show()
```
